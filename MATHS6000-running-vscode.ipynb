# AppsAnywhere (but not your settings)

I'm reasonably sure the way `AppsAnywhere` program works is to open your application _in the cloud_, but unfortunately your settings etc are stored locally. This means **every single** time you open `vscode` in this way, you will need to reset the default interpretter and install python extensions. I don't currently know of any solution to this.
