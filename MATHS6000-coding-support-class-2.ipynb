## If Statements

We can use `if`, `elif`, and  `else` to control flow through the program.

```python
i=int(input("Input a value of i"))

if i<0:
    print(i," is negative")
elif i==0:
    print(i," is zero")
else:
    print(i," is positive")
```

To execute more than one statement for an `if` condition, indent all statements that belong to that block.

```python
i=int(input("Input a value of i"))

if i<0:
    i = i*100
    print(i," is negative")
elif i==0:
    print(i," is zero")
else:
    i = i*200
    print(i," is positive")
```

## For loops

A for loop repeats a block of code over a sequence of values. This is valuable as there are hardly any situations in numerics where you wouldn't want to put a limit on the number of times you would want to do something unlike other common tasks.

The syntax in python is
~~~
for <iterator> in <collection>:
    <loop body>
~~~
and the most common usage for us will be of the form
~~~
for i in range(10):
~~~
where this will iterate the loop 10 times, starting with `i=0` and ending with `i=9`.
    

```python
for i in range(10):
    print(" Execute the loop, i:=",i)
```

Remember many algorithms in mathematics can be written in a form like
$$
A = \sum_{i=0}^n f(i)
$$
which can be replicated with the code
~~~
total=0.
for i in range(n+1):
    total = total + f(i)
A = total
~~~
You can easily change the range by giving a start and finish value, for example
~~~
range(1,n) # goes from 1,2,...,n-1
~~~
This `range(start, stop)` generates integer values from `start` up to, but not including, `stop`.

## Exiting the Loop

As stated above we will normally want to limit the maximum number of times we do something, but we may also want to finish early if a condition  is met, i.e. a solution has been found. You can break from a loop with the `break` command. Say we want to calculate the infinite sum
$$
S = \sum_{i=1}^\infty \frac{1}{2^i + 1},
$$
we know this is a convergent series. Starting at $i=1$ each successive term is smaller than the last, so to check for convergence we can stop when
$$
\frac{1}{2^i+1} < \epsilon
$$
where $\epsilon>0$ is our tolerance.

```python
n=10000
total=0.
tol = 1.e-6
for i in range(1,n+1):
    term = 1.0 / (2**i + 1)
    total = total + term
    if term < tol:
        print(" Exit after ",i," loops, S:=",total)
        break
```

# Functions

The general syntax for a function is:
~~~
def <function_name>(<arguments>):
    <function statements>
~~~
A function definition must be executed before the function is called.

```python
## Example Function
def square(x):
    return x*x

i = 3
print( square(i) )
```

We can define multiple functions, call one function from another, and combine them to build more complicated calculations. A function can return a value, which can then be used in another expression or passed as an argument to another function.

```python
def square(x):
    return x*x

def add_one(x):
    return x + 1

def square_then_add_one(x):
    return add_one(square(x))

print(square_then_add_one(3))
```

In particular, note that functions themselves can be passed as arguments.

```python
def apply_twice(f, x):
    return f(f(x))

print(apply_twice(square, 2))
```

## File Output

- There may be times when you want to output data to a file, for example results for a table.
- The command to output results is quite simple, however often we want data to be formatted (you might only want to print the first 8 digits of a float rather than all 16).
- When writing formatted text to a file, we usually convert values into a string representation.
- Normally you want to create a line of text in a string, then write the string to file, so the code looks something like:
~~~
with open("MyResults.csv", "w") as f:
    ...
    n = ...                     # some number
    result = ...                # some result
    line = f" {n} , {result} \n"  # put n in first, separate by comma, put the result, then end the line with '\n'
    f.write(line)                # write to file
    ...
# file is automatically closed when block finishes
print("File write successful.")
~~~
- Some objects in python already have automatic ways to write into csv format, for example `numpy`, please look them up before writing yourself
- See the [Python Documentation on strings](https://docs.python.org/3/library/string.html#formatstrings) for more details on formatting
- Although this website: [https://pyformat.info/](https://pyformat.info/) gives more practical explanations that are a bit easier to understand.

```python
with open("MyResults.csv", "w") as f:
    n_vals=[10 , 20 , 40, 80,160]
    for n in n_vals:
        result = 1/(n*2+1)
        line = f" {n:5d} , {result:12.8f} \n"
        f.write(line)
print("File written")
```
