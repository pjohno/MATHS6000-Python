# Running Python Files on Spyder (inside Anaconda)

If the machine you are working on has anaconda installed, we can open the application `spyder` from inside it to run code. Click on the windows start menu and open `Anaconda Navigator`.

# IMPORTANT STEP

If you want to save your work and be able to access it from different PCs around the university (and at home), you must setup the working directory to be on your onedrive folder.

Select to save the file from the menu. Click on the onedrive folder (mine is called "Paul - The University of Manchester" and it has a link in the left hand bar). Create a new directory to save your work in, I'm going to call it `MATH60000`.

<div style='text-align: left;'>
<img src="images/2026/Screenshot131224.png" alt="image" width="50%" height="auto">
</div>

Now save the file as `helloWorld.py`. Use the `####.py` ending to tell the computer this is a python file.
<div style='text-align: left;'>
<img src="images/2026/Screenshot131251.png" alt="image" width="30%" height="auto">
</div>

Run the code again to check everything is working. You should see that working directory (on the top right) is listed as your onedrive folder. If you open `spyder` on another machine and want to work on this code again, make sure you change the working directory to the correct one.
